```python
# pip install requests beautifulsoup4
import requests
from bs4 import BeautifulSoup

html = requests.get("https://talks.contact.ms/").text
soup = BeautifulSoup(html, "html.parser")

slides = soup.select("section") or soup.select("[class*=slide]")
for n, s in enumerate(slides, 1):
    print(f"--- slide {n} ---")
    print(s.get_text("\n", strip=True))
```

```python
# %pip install playwright img2pdf && playwright install chromium
import asyncio, img2pdf
from playwright.async_api import async_playwright

async def main():
    shots, prev = [], None
    async with async_playwright() as p:
        browser = await p.chromium.launch()
        page = await browser.new_page(viewport={"width": 1920, "height": 1080})
        await page.goto("https://talks.contact.ms/#1")
        await page.wait_for_load_state("networkidle")
        for i in range(200):
            await page.wait_for_timeout(700)   # let transitions finish
            png = await page.screenshot()
            if png == prev:                    # nothing changed, so we're at the end
                break
            path = f"slide_{i+1:02d}.png"
            open(path, "wb").write(png)
            shots.append(path)
            prev = png
            await page.keyboard.press("ArrowRight")
        await browser.close()
    open("talk.pdf", "wb").write(img2pdf.convert(shots))

await main()
```

```python
import asyncio
from playwright.async_api import async_playwright

async def main():
    async with async_playwright() as p:
        b = await p.chromium.launch()
        page = await b.new_page()
        await page.goto("https://talks.contact.ms/")
        await page.wait_for_load_state("networkidle")
        await page.pdf(path="talk_text.pdf", width="1920px", height="1080px",
                       print_background=True)
        await b.close()

await main()
```
